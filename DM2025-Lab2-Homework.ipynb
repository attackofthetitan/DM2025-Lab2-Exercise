## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

I first observed that the data was very sanitized, and that the data was mostly in the form of reddit/twitter posts.

The raw data comes from multiple files (`data_identification.csv`, `emotion.csv`, `final_posts.json`) and must first be merged. I use the `id` field as the key to join:

- `data_identification.csv` → provides the `split` column (`train` / `test`).
- `emotion.csv` → provides the target label `emotion` for training samples.
- `final_posts.json` → contains the actual tweet text and hashtags under the nested `post` field.

After merging, I keep only rows with non-missing labels for training and use the `split` column to separate train/test.

For text preprocessing I implement a `normalize_text()` function with the following steps:

- **HTML & whitespace normalization**
  - Decode HTML with `html.unescape`.
  - Replace line breaks, carriage returns, and tabs with spaces.
  - Collapse multiple spaces into a single space and strip leading/trailing spaces.

- **Emoji handling**
  - Convert emojis to textual descriptions using `emoji.demojize` in English.
  - Remove the colon characters around emoji names so they behave like normal tokens.

- **Special placeholders**
  - Replace user mentions like `@username` with the token `<USER>`.
  - Replace URLs (`http...` or `www...`) with `<URL>`.
  - Replace `[NAME]` placeholders with `<NAME>`.
  - These placeholders are **added to the tokenizer vocabulary** so the model can learn specific embeddings for them.

- **Quote & “edit” cleanup**
  - Remove leading quote markers (`>` at the start of the text). Usually appears in green text form for story telling.
  - If the text contains “edit:” or “tl;dr” (case-insensitive), drop everything after that, keeping only the original main content.

- **Hashtag integration**
  - If hashtag metadata is available as a list, I append them to the end of the text in the form `#tag1 #tag2 ...`.
  - This keeps information that might not be explicitly present in the tweet body.

This preprocessing is applied consistently to both the training and the test splits, so the model sees text in a normalized, consistent format while preserving important semantic cues such as emojis, mentions, URLs, and hashtags.

---

### 1.2 Feature Engineering Steps

I do not create hand-crafted numerical features, I rely on the transformer’s contextual embeddings. However, I still perform some feature engineering designs inside the text representation:

- **Special tokens as features**
  - `<USER>`, `<URL>`, and `<NAME>` are treated as dedicated tokens and added to the tokenizer vocabulary.
  - This allows the model to learn a consistent representation for user mentions and URLs instead of treating them as many different rare tokens.

- **Emojis as lexical features**
  - By converting emojis into descriptive words (e.g., `:smiling_face:` → `smiling face`), I expose emotional signals directly to the model as text tokens, which is especially relevant for emotion recognition.

- **Hashtags as extra context**
  - Hashtags often encode emotion or topic. Concatenating them to the text effectively turns them into additional tokens the model can learn from.

- **Tokenization**
  - I use the HuggingFace `AutoTokenizer` corresponding to each backbone model.
  - Texts are tokenized with truncation and a fixed maximum length of 160-192 tokens because the max length of content was 89 words, or 440 characters. Longer length was not needed.
  - Padding is done to a fixed length, and training uses standard attention masks.

Overall, the “features” are mainly the tokenized texts enriched with normalized emojis, hashtags, and special placeholders, which the transformer backbone converts into high-dimensional representations.

### 1.3 Explanation of Your Model

#### Backbone models

I fine-tune transformer-based models that are already pre-trained on emotion-related or Twitter data:

- `Amirhossein75/multi-label-emotion-classification-reddit-comments-roberta`
- `cardiffnlp/twitter-roberta-large-emotion-latest`

Both are RoBERTa-style models specialized for emotion classification or social-media text, which aligns well with this task. I train each backbone with **5-fold Stratified K-Fold cross-validation** to obtain out-of-fold (OOF) predictions and more robust estimates.

#### Custom classification head

Instead of the default single linear layer, I replace the head with a **Multi-Sample Dropout (MSD)** head:

- Take the `[CLS]` (or first token) hidden state from the backbone.
- Apply several independent dropout layers (e.g., 5 samples).
- Pass each dropped version through the same linear classifier.
- Average the logits from all samples to get the final prediction.

This acts as an inexpensive ensemble at the head level and improves regularization.

#### Loss function, class imbalance, and regularization

To handle class imbalance and improve generalization, I use a custom `WeightedTrainer`:

- **Class weights**
  - Compute `class_weight` using `sklearn`’s `compute_class_weight("balanced", ...)`.
  - Optionally “smooth” them by raising to a power (e.g., 0.5) so that weights are not too extreme.
  - Pass these weights into the loss function so rare emotions get higher penalty when misclassified.

- **Label smoothing**
  - Apply a small label smoothing factor (e.g., 0.05) in the cross-entropy loss to reduce overconfidence and improve calibration.

- **R-Drop (consistency regularization)**
  - For each batch during training, I run the model **twice** with dropout on the same inputs.
  - Compute:
    - Cross-entropy loss for each forward pass.
    - A symmetric KL divergence between the two predicted distributions.
  - The final loss is:  
    `L = 0.5 * (CE1 + CE2) + α * KL(p1 || p2)`  
    where α is a weight (`rdrop_alpha`).
  - This forces the model to produce consistent predictions under dropout noise and improves robustness.

- *(Focal loss is implemented and could be turned on via a `gamma` parameter, but in the final configuration the standard (weighted) cross-entropy with label smoothing is used.)*

#### Optimization details

Key training hyper-parameters:

- Optimizer: `AdamW` with weight decay.
- Learning rates per model (e.g., ~1.5e-5) chosen from small values suitable for large transformers.
- Scheduler: cosine learning-rate schedule with warmup (`warmup_ratio = 0.1`).
- Gradient clipping (`max_grad_norm = 0.5`).
- Batch size: 16 for training; larger for evaluation.
- Mixed precision: use `fp16` or `bf16` when available on the GPU to speed up training.

I also implement **Layer-Wise Learning Rate Decay (LLRD)**:

- Parameters in lower (earlier) transformer layers receive a smaller learning rate than higher layers.
- The decay factor is controlled by `layerwise_lr_decay` (e.g., 0.95).
- This keeps the bottom layers closer to the pre-trained weights and lets the top layers adapt more to the emotion task.

Training uses **early stopping** and `load_best_model_at_end=True` based on validation F1 to avoid overfitting.

#### Ensembling

After training all folds and backbones, I generate OOF predictions and test predictions:

1. **Base models**
   - For each backbone and each fold, I get:
     - OOF logits for the training data.
     - Logits for the test data.

2. **Probability calibration**
   - Apply `softmax` to obtain class probabilities.

3. **Two types of ensembles**
   - **Simple average**: average probabilities across all models.
   - **Stacking with Logistic Regression**:
     - Concatenate all model OOF probabilities as features.
     - Train a multinomial Logistic Regression meta-model on the OOF predictions.
     - Apply this meta-model to the concatenated test probabilities.

4. **Choosing the final ensemble**
   - Compare macro / micro F1 of the average vs. stacked ensemble on OOF data.
   - Use the one with higher macro-F1 as the final ensemble for predictions.

#### Pseudo-labeling

To leverage unlabeled test data, I perform **pseudo-labeling** in a conservative way:

1. Use the initial ensemble trained only on the gold training set to predict probabilities for test samples.
2. Compute:
   - **Confidence**: max probability over classes.
   - **Entropy**: `-Σ p * log p` to measure uncertainty.
3. Select test samples that are:
   - high confidence (`confidence > 0.85`), and  
   - low entropy (`entropy < 0.30`).
4. Assign pseudo-labels to those samples and mark them as additional training data, but cap the size:
   - Use at most **50% of the original training size** as pseudo-labeled data to avoid overwhelming the real labels.
5. Retrain the models (same backbones and setup) on the expanded dataset and build a new ensemble.
6. Evaluate on the original training set (via OOF predictions) and use the improved ensemble to generate the final submission file.

This two-stage training (base ensemble → pseudo-labelled retraining) aims to squeeze extra performance from confident unlabeled examples while trying to remain safe with strict thresholds and a size cap.

### 2.2 Mention Insights You Gained

Some key insights from this project:

- **Domain-specific pretraining matters**
  - Models already trained on emotions or social-media data (Reddit/Twitter) are strong starting points for this competition and generally easier to fine-tune than generic language models.

- **Good preprocessing helps transformers**
  - Even though transformers can handle noisy text, making emojis, hashtags, and special tokens explicit gave the model clearer emotional cues and improved interpretability of the inputs.

- **Regularization and small tweaks add up**

  - Pseudo-labels can improve performance, but only when the model is very confident and the amount of pseudo data is controlled. Otherwise, noisy labels can hurt more than help.

## 2. Feature Engineering Steps

```python
MULTI_WS    = re.compile(r"\s+")
LEAD_QUOTE  = re.compile(r"^\s*>+\s*")
NAME_PLACE = re.compile(r"\[NAME\]", re.IGNORECASE)
URL_PATTERN   = re.compile(r"http\S+|www\.\S+")
USER_PATTERN  = re.compile(r"@\w+")
EDIT_PATTERN = re.compile(r"(edit:|tl;dr)", re.IGNORECASE)

def normalize_text(text: str, hashtags: Optional[str] = None) -> str:
    text = html.unescape(text)
    text = text.replace("\r", " ").replace("\n", " ").replace("\t", " ")

    # demojize
    text = emoji.demojize(text, language="en").replace(":", " ")
    
    # normalize user mentions, urls, name placeholders
    text = USER_PATTERN.sub(" <USER> ", text)
    text = URL_PATTERN.sub(" <URL> ", text)
    text = NAME_PLACE.sub(" <NAME> ", text)

    # remove leading > quotes
    text = LEAD_QUOTE.sub("", text)

    m = EDIT_PATTERN.search(text)
    if m:
        text = text[:m.start()]
        
    if isinstance(hashtags, list) and len(hashtags) > 0:
        tag_tokens = " ".join(f"#{h}" for h in hashtags if isinstance(h, str))
        text = f"{text} {tag_tokens}"

    # cleanup whitespace
    text = MULTI_WS.sub(" ", text).strip()
    return text

train_df = df[df["split"]=="train"].dropna(subset=["emotion"]).reset_index(drop=True)
test_df  = df[df["split"]=="test"].reset_index(drop=True)

train_df["text"] = train_df.apply(lambda r: normalize_text(r["text"], r["hashtags"]), axis=1)
test_df["text"] = test_df.apply(lambda r: normalize_text(r["text"], r["hashtags"]), axis=1)

labels = sorted(train_df["emotion"].unique().tolist())
label2id = {l:i for i,l in enumerate(labels)}
id2label = {i:l for l,i in label2id.items()}

y = train_df["emotion"].map(label2id).values.astype(int)
print("Train:", train_df.shape, "| Test:", test_df.shape, "| Classes:", labels)
```

```python
# helpers
def make_ds(texts, labels=None):
    d = {"text": texts}
    if labels is not None:
        d["label"] = labels
    return Dataset.from_dict(d)

def tokenize_fn(examples, tokenizer, max_len=256):
    return tokenizer(examples["text"], truncation=True, max_length=max_len, padding="max_length",)

def compute_f1(eval_pred):
    logits = eval_pred.predictions
    labels = eval_pred.label_ids

    # logits = logits[0]
    logits = np.asarray(logits)
    preds = logits.argmax(axis=-1)

    macro = f1_score(labels, preds, average="macro")
    micro = f1_score(labels, preds, average="micro")
    return {"f1_macro": macro, "f1_micro": micro}

def augment_text(text: str) -> str:
    words = text.split()
    if len(words) <= 4:
        return text
    keep = []
    for w in words:
        # drop 10% of tokens at random
        if np.random.rand() < 0.1:
            continue
        keep.append(w)
    if not keep:
        keep = words
    return " ".join(keep)
    
def get_layer(name: str, num_hidden_layers: int) -> int:
    if "embeddings" in name:
        return 0
    for i in range(num_hidden_layers):
        if f"encoder.layer.{i}." in name:
            return i + 1
    return num_hidden_layers + 1
```

## 3. Model Implementation Steps

```python
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, weight=None, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.gamma = gamma
        self.weight = weight
        self.reduction = reduction

    def forward(self, inputs, targets):
        ce_loss = F.cross_entropy(inputs, targets, weight=self.weight, reduction='none')
        pt = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        if self.reduction == 'mean':
            return focal_loss.mean()
        return focal_loss.sum()

# weighted trainer with r-drop, llrd, label smoothing, focal loss, all activated through init args
class WeightedTrainer(Trainer):
    def __init__(self,class_weight=None, rdrop_alpha: float = 0.0, layerwise_lr_decay: float = 1.0, 
                    label_smoothing_factor: float = 0.05, gamma: float = 0.0, **kwargs,):
        super().__init__(**kwargs)
        class_weight = torch.tensor(class_weight, dtype=torch.float)
        self.class_weight = class_weight
        self.gamma = float(gamma)
        self.rdrop_alpha = float(rdrop_alpha)
        self.layerwise_lr_decay = float(layerwise_lr_decay)
        self.label_smoothing_factor = float(label_smoothing_factor)
        
    # llrd
    def create_optimizer(self):
        if self.optimizer is not None:
            return self.optimizer

        lr = self.args.learning_rate
        wd = self.args.weight_decay
        layer_decay = self.layerwise_lr_decay

        if layer_decay == 1.0:
            return super().create_optimizer()

        model = self.model
        num_layers = getattr(model.config, "num_hidden_layers", 12)

        layer_parameters = {}
        for name, param in model.named_parameters():
            if not param.requires_grad:
                continue
            layer_id = get_layer(name, num_layers)
            layer_parameters.setdefault(layer_id, []).append((name, param))

        optimizer_grouped_parameters = []
        max_layer_id = max(layer_parameters.keys())
        no_decay = ["bias", "LayerNorm.weight"]

        for layer_id in sorted(layer_parameters.keys()):
            scale = layer_decay ** (max_layer_id - layer_id)
            layer_lr = lr * scale
            layer_params = layer_parameters[layer_id]

            decay_params = [
                p for n, p in layer_params
                if not any(nd in n for nd in no_decay)
            ]
            nodecay_params = [
                p for n, p in layer_params
                if any(nd in n for nd in no_decay)
            ]

            if decay_params:
                optimizer_grouped_parameters.append({"params": decay_params, "lr": layer_lr, "weight_decay": wd})
            if nodecay_params:
                optimizer_grouped_parameters.append(
                    {"params": nodecay_params, "lr": layer_lr, "weight_decay": 0.0}
                )

        self.optimizer = AdamW(
            optimizer_grouped_parameters,
            lr=lr,
            betas=(self.args.adam_beta1, self.args.adam_beta2),
            eps=self.args.adam_epsilon,
        )
        return self.optimizer

    def ce_loss(self, logits, labels):
        if labels.ndim > 1:
            labels = labels.view(-1)
        labels = labels.long()

        if logits.ndim > 2:
            logits = logits.view(logits.size(0), -1)

        assert logits.size(0) == labels.size(0), (f"logits batch {logits.size()} vs labels {labels.size()}")

        weight = self.class_weight.to(logits.device)
        gamma = self.gamma
        label_smoothing = self.label_smoothing_factor
        if gamma > 0.0:
            focal_loss_fn = FocalLoss(gamma=gamma, weight=weight)
            return focal_loss_fn(logits, labels)
        else:
            return F.cross_entropy(logits, labels, weight=weight, label_smoothing=label_smoothing)

    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.pop("labels")
        outputs1 = model(**inputs)
        logits1 = outputs1.logits

        if self.rdrop_alpha > 0.0:
            outputs2 = model(**inputs)
            logits2 = outputs2.logits

            loss1 = self.ce_loss(logits1, labels)
            loss2 = self.ce_loss(logits2, labels)
            loss_ce = 0.5 * (loss1 + loss2)

            # symmetric kl on softmax
            p_log = F.log_softmax(logits1, dim=-1)
            q_log = F.log_softmax(logits2, dim=-1)
            p = p_log.exp()
            q = q_log.exp()
            kl_pq = F.kl_div(p_log, q, reduction="batchmean")
            kl_qp = F.kl_div(q_log, p, reduction="batchmean")
            loss_kl = 0.5 * (kl_pq + kl_qp)

            loss = loss_ce + self.rdrop_alpha * loss_kl

            if return_outputs:
                avg_logits = 0.5 * (logits1 + logits2)
                outputs1.logits = avg_logits
                return loss, outputs1
            return loss
        else:
            loss = self.ce_loss(logits1, labels)
            if return_outputs:
                return loss, outputs1
            return loss
```

```python
class MultiHeadClassifier(nn.Module):
    def __init__(self, input_dim, num_labels, n_heads=8, p=0.2):
        super().__init__()
        self.n_heads = n_heads
        
        # intermediate layer for each head
        self.intermediates = nn.ModuleList([
            nn.Sequential(
                nn.Linear(input_dim, input_dim // 2),
                nn.GELU(),
                nn.Dropout(p)
            ) for _ in range(n_heads)
        ])
        
        self.classifiers = nn.ModuleList([
            nn.Linear(input_dim // 2, num_labels) 
            for _ in range(n_heads)
        ])
        
        # learnable weights for heads
        self.head_weights = nn.Parameter(torch.ones(n_heads) / n_heads)
        
        for classifier in self.classifiers:
            nn.init.normal_(classifier.weight, mean=0.0, std=0.02)
            nn.init.zeros_(classifier.bias)

    def forward(self, x):
        if x.dim() == 3:
            x = x[:, 0, :]  # [B, H]
        
        # combine heads weighted
        weights = F.softmax(self.head_weights, dim=0)
        logits = 0
        
        for i in range(self.n_heads):
            hidden = self.intermediates[i](x)
            head_logits = self.classifiers[i](hidden)
            logits = logits + weights[i] * head_logits
        
        return logits

# multi-sample dropout head
class MSDHead(nn.Module):
    def __init__(self, input_dim: int, num_labels: int, num_samples: int = 5, p: float = 0.3):
        super().__init__()
        self.num_samples = num_samples
        self.dropouts = nn.ModuleList([nn.Dropout(p) for _ in range(num_samples)])
        self.fc = nn.Linear(input_dim, num_labels)

        nn.init.normal_(self.fc.weight, mean=0.0, std=0.02)
        nn.init.zeros_(self.fc.bias)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        if x.dim() == 3:
            x = x[:, 0, :]

        logits_sum = 0
        for drop in self.dropouts:
            logits_sum = logits_sum + self.fc(drop(x))

        return logits_sum / self.num_samples

```

```python
def train_folds(model_name, n_splits=5, lr=2e-5, epochs=3, bsz=16, max_len=192, wd=0.01, seed=SEED, n_heads=4, head_dropout=None, multihead=False, msd_samples=5):

    tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)

    new_tokens = ["<USER>", "<URL>", "<NAME>"]
    new_tokens = [t for t in new_tokens if t not in tokenizer.get_vocab()]
    
    if new_tokens:
        print(f"Adding {len(new_tokens)} special tokens: {new_tokens}")
        tokenizer.add_tokens(new_tokens)
    
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)

    oof_preds = np.zeros((len(train_df), len(labels)), dtype=np.float32)
    test_logits_accum = np.zeros((len(test_df), len(labels)), dtype=np.float32)

    for fold, (tr_idx, va_idx) in enumerate(skf.split(train_df, y), 1):

        output_dir = f"./out_{os.path.basename(model_name)}_f{fold}_{seed}"
        print(f"\n[{model_name}] Fold {fold}/{n_splits}")
        tr_df = train_df.iloc[tr_idx].copy()

        tr_texts = tr_df["text"].tolist()
        tr_labels = tr_df["emotion"].map(label2id).tolist()

        va_texts = train_df.loc[va_idx, "text"].tolist()
        va_labels = train_df.loc[va_idx, "emotion"].map(label2id).tolist()

        ds_tr = make_ds(tr_texts, tr_labels).map(lambda ex: tokenize_fn(ex, tokenizer, max_len), batched=True)
        ds_va = make_ds(va_texts, va_labels).map(lambda ex: tokenize_fn(ex, tokenizer, max_len), batched=True)

        tr_y = y[tr_idx]
        classes = np.unique(tr_y)
        raw_class_weight = compute_class_weight("balanced", classes=classes, y=tr_y)
        gamma = 0.5
        class_weight = np.power(raw_class_weight, gamma)

        ds_te = make_ds(test_df["text"].tolist()).map(lambda ex: tokenize_fn(ex, tokenizer, max_len), batched=True)

        cols = ["input_ids","attention_mask","label"]
        ds_tr.set_format(type="torch", columns=[c for c in cols if c in ds_tr.column_names])
        ds_va.set_format(type="torch", columns=[c for c in cols if c in ds_va.column_names])
        ds_te.set_format(type="torch", columns=[c for c in ["input_ids","attention_mask"] if c in ds_te.column_names])

        config = AutoConfig.from_pretrained(
            model_name,
            num_labels=len(labels),
            label2id=label2id,
            id2label=id2label,
            hidden_dropout_prob=0.2,
            attention_probs_dropout_prob=0.2,
        )

        config.problem_type = "single_label_classification"

        model = AutoModelForSequenceClassification.from_pretrained(
            model_name,
            config=config,
            ignore_mismatched_sizes=True,
        )
        
        hidden_size = config.hidden_size
        
        if head_dropout is None:
            head_dropout = config.hidden_dropout_prob
        
        model.resize_token_embeddings(len(tokenizer))     

        if multihead:
            model.classifier = MultiHeadClassifier(
                input_dim=config.hidden_size,
                num_labels=len(labels),
                n_heads=n_heads,
                p=head_dropout,
            )
        else:
            model.classifier = MSDHead(
                input_dim=config.hidden_size,
                num_labels=len(labels),
                num_samples=msd_samples,
                p=head_dropout,
            )
           
        args = TrainingArguments(
            output_dir=output_dir,
            learning_rate=lr,
            per_device_train_batch_size=bsz,
            per_device_eval_batch_size=bsz*2,
            num_train_epochs=epochs,
            weight_decay=wd,
            eval_strategy="epoch",
            save_strategy="epoch",
            load_best_model_at_end=True,
            metric_for_best_model="f1_micro", 
            save_total_limit=1,
            logging_steps=50,
            greater_is_better=True,
            report_to="none",
            seed=seed+fold,
            fp16=use_fp16 and not use_bf16,
            bf16=use_bf16,
            lr_scheduler_type="cosine",
            warmup_ratio=0.1,
            max_grad_norm=0.5,
            optim="adamw_torch",
        )

        data_collator = default_data_collator 
        trainer = WeightedTrainer(
            class_weight=class_weight,
            rdrop_alpha=1.0,
            layerwise_lr_decay=0.95,
            label_smoothing_factor=0.05,
            model=model,
            gamma=0.0,
            args=args,
            train_dataset=ds_tr,
            eval_dataset=ds_va,
            processing_class=tokenizer,
            data_collator=data_collator,
            compute_metrics=compute_f1,
            callbacks=[EarlyStoppingCallback(early_stopping_patience=2)]
        )

        resume_ckpt = None
        if os.path.isdir(output_dir):
            ckpts = glob.glob(os.path.join(output_dir, "checkpoint-*"))
            if ckpts:
                ckpts_sorted = sorted(ckpts, key=lambda p: int(p.split("-")[-1]))
                resume_ckpt = ckpts_sorted[-1]

        if resume_ckpt is not None:
            print("Resuming from checkpoint:", resume_ckpt)
            trainer.train(resume_from_checkpoint=resume_ckpt)
        else:
            print("No checkpoint found, retraining")
            trainer.train()

        valid_logits = trainer.predict(ds_va).predictions
        oof_preds[va_idx] = valid_logits

        test_logits = trainer.predict(ds_te).predictions
        test_logits_accum += test_logits / n_splits

        va_pred = np.argmax(valid_logits, axis=1)
        f1 = f1_score(va_labels, va_pred, average="macro")

        print(f"Fold {fold} F1(macro): {f1:.4f}")

        del trainer, model
        torch.cuda.empty_cache()

    oof_pred_labels = oof_preds.argmax(axis=1)
    oof_f1_macro = f1_score(y, oof_pred_labels, average="macro")
    oof_f1_micro = f1_score(y, oof_pred_labels, average="micro")
    print(f"\n[Model {model_name}] OOF F1(macro): {oof_f1_macro:.4f} | micro: {oof_f1_micro:.4f}")

    return oof_preds, test_logits_accum
```

```python
torch.cuda.empty_cache()

SEEDS = [42] 

backbones = [
    # "GroNLP/hateBERT",
    # "microsoft/deberta-v3-large",
    # "FacebookAI/roberta-large",
    # "j-hartmann/emotion-english-roberta-large",
    "Amirhossein75/multi-label-emotion-classification-reddit-comments-roberta",
    "cardiffnlp/twitter-roberta-large-emotion-latest",
]
cfgs = {
    "j-hartmann/emotion-english-roberta-large": dict(lr=1e-5, epochs=10),
    "FacebookAI/roberta-large": dict(lr=1.5e-5, epochs=4),
    "microsoft/deberta-v3-large": dict(lr=8e-6, epochs=4),
    "Amirhossein75/multi-label-emotion-classification-reddit-comments-roberta": dict(lr=1.5e-5, epochs=4),
    "GroNLP/hateBERT": dict(lr=1.5e-5, epochs=10),
    "cardiffnlp/twitter-roberta-large-emotion-latest": dict(lr=1.5e-5, epochs=4),
}

all_oof, all_test = [], []

for name in backbones:
    hp = cfgs[name]
    for s in SEEDS:
        print(f"\n {name} | SEED {s} ")
        oof, test_logits = train_folds(
            name,
            n_splits=5,
            lr=hp["lr"],
            epochs=hp["epochs"],
            bsz=16,
            max_len=160,
            wd=0.01,
            seed=s,
            head_dropout=0.1,
        )
        all_oof.append(softmax(oof, axis=1))
        all_test.append(softmax(test_logits, axis=1))
```

```python
def run_ensemble(all_oof, all_test, y_true, desc=""):
    # lr stacking
    X_oof = np.hstack(all_oof)
    meta_model = LogisticRegression(
        multi_class="multinomial",
        max_iter=1000,
        C=0.1,
        random_state=SEED,
    )
    meta_model.fit(X_oof, y_true)

    oof_meta_probs = meta_model.predict_proba(X_oof)
    oof_meta_pred = oof_meta_probs.argmax(axis=1)
    f1_meta_macro = f1_score(y_true, oof_meta_pred, average="macro")
    f1_meta_micro = f1_score(y_true, oof_meta_pred, average="micro")

    print(f"[Stacked]  macro={f1_meta_macro:.4f}, micro={f1_meta_micro:.4f}")

    # average
    oof_avg = np.mean(all_oof, axis=0)
    oof_avg_pred = oof_avg.argmax(axis=1)
    f1_avg_macro = f1_score(y_true, oof_avg_pred, average="macro")
    f1_avg_micro = f1_score(y_true, oof_avg_pred, average="micro")

    print(f"[Average]  macro={f1_avg_macro:.4f}, micro={f1_avg_micro:.4f}")

    # choose ensemble by macro f1
    if f1_meta_macro > f1_avg_macro:
        print("-> Using stacked ensemble")
        oof_best = oof_meta_probs
        X_test = np.hstack(all_test)
        test_best = meta_model.predict_proba(X_test)
    else:
        print("-> Using average ensemble")
        oof_best = oof_avg
        test_best = np.mean(all_test, axis=0)

    pred_best = oof_best.argmax(axis=1)
    f1_best_macro = f1_score(y_true, pred_best, average="macro")
    f1_best_micro = f1_score(y_true, pred_best, average="micro")
    print(f"[Chosen]   macro={f1_best_macro:.4f}, micro={f1_best_micro:.4f}")

    return oof_best, test_best


pre_oof_ens, pre_test_ens = run_ensemble(all_oof, all_test, y, desc="Pre-pseudo")

pre_pred = pre_oof_ens.argmax(axis=1)
pre_f1_macro = f1_score(y, pre_pred, average="macro")
pre_f1_micro = f1_score(y, pre_pred, average="micro")

print("\n[Pre-pseudo ensemble] Classification report:")
print(classification_report(y, pre_pred, target_names=labels))
print(f"[Pre-pseudo ensemble] macro={pre_f1_macro:.4f}, micro={pre_f1_micro:.4f}")

cm = confusion_matrix(y, pre_pred)
print("\n[Pre-pseudo ensemble] Confusion matrix:")
print(cm)


pre_labels = [id2label[i] for i in pre_test_ens.argmax(axis=1)]
submission_pre = pd.DataFrame({"id": test_df["id"].values, "emotion": pre_labels})
os.makedirs("./submit", exist_ok=True)
pre_path = "./submit/submission_pre_pseudo_ensemble.csv"
submission_pre.to_csv(pre_path, index=False)
print("Saved pre-pseudo ensemble submission:", pre_path)
```
